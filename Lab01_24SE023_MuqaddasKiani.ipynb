```python
question = 'Q7: Grap the domain from "user@domain.com".'

def get_domain(email):
    return email.split("@")[1]

email = "user@domain.com"
answer = get_domain(email)

print(question)
print("Answer:", answer)
```

```python
#CASE INSENSITIVE
question = 'Q8: Write a function to check if "dog" is in a string.'

def has_dog(text):
    return "dog" in text.lower()

sentence = "The DOG is playing."
answer = has_dog(sentence)

print(question)
print("Answer:", answer)
```

```python
question = 'Q9: Count how many times "dog" appears in a string.'

def count_dog(text):
    words = text.lower().split()
    return words.count("dog")

sentence = "Dog runs, dog sleeps, and dog plays."
answer = count_dog(sentence)

print(question)
print("Answer:", answer)
```

```python
question = "Q10: Check speeding ticket categories."

def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed = speed - 5

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"

print(question)
print("Answer 1:", caught_speeding(60, False))
print("Answer 2:", caught_speeding(77, False))
print("Answer 3:", caught_speeding(83, False))
print("Answer 4:", caught_speeding(85, True))
```
